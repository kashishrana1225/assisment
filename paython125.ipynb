```python
# ============================================================
# 📊 ECONOMIC & FINANCIAL DATA EDA (FULLY WORKING SCRIPT)
# ============================================================

# ✅ Step 1: Import all required libraries
import pandas as pd
import numpy as np

# ✅ Step 2: Load your dataset
# ⚠️ Make sure your file name matches or update the path accordingly
file_path = "/content/finance_economics_dataset.csv.crdownload"   # <-- change if needed
df = pd.read_csv(file_path)

# ✅ Step 3: Display basic info
print("===== BASIC DATA INFO =====")
print("Shape of dataset:", df.shape)
print("\nColumn Names:\n", list(df.columns))
print("\nData Types:\n")
print(df.dtypes)
print("=" * 50)

# ✅ Step 4: Convert 'Date' column to datetime (if exists)
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# ✅ Step 5: Numeric data only
numeric_df = df.select_dtypes(include=[np.number])

# ============================================================
# 🔍 QUESTIONS 1–20 ANSWERS
# ============================================================

# 1️⃣ Shape
print("\n1️⃣ Shape of dataset:", df.shape)

# 2️⃣ Column names and types
print("\n2️⃣ Columns and data types:\n", df.dtypes)

# 3️⃣ Unique stock indices
if 'Stock Index' in df.columns:
    print("\n3️⃣ Unique Stock Indices:", df['Stock Index'].nunique())
else:
    print("\n3️⃣ 'Stock Index' column not found!")

# 4️⃣ Date range
if 'Date' in df.columns:
    print("\n4️⃣ Date Range:", df['Date'].min(), "to", df['Date'].max())

# 5️⃣ Missing values
print("\n5️⃣ Missing Values:\n", df.isnull().sum())

# 6️⃣ Negative values
print("\n6️⃣ Negative Value Counts (numeric columns):\n", (numeric_df < 0).sum())

# 7️⃣ GDP Growth summary
if 'GDP Growth (%)' in df.columns:
    print("\n7️⃣ GDP Growth Summary:\n", df['GDP Growth (%)'].describe())

# 8️⃣ Zero or near-zero trading volume
if 'Trading Volume' in df.columns:
    print("\n8️⃣ Zero or near-zero Trading Volume rows:",
          (df['Trading Volume'] <= 1).sum())

# 9️⃣ Duplicate rows
print("\n9️⃣ Duplicate Rows:", df.duplicated().sum())

# 🔟 Outliers in GDP, Gold, or Oil using IQR
print("\n🔟 Outliers (GDP, Gold, Oil):")
for col in ['GDP Growth (%)', 'Gold Price (USD per Ounce)', 'Crude Oil Price (USD per Barrel)']:
    if col in df.columns:
        Q1, Q3 = df[col].quantile([0.25, 0.75])
        IQR = Q3 - Q1
        outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
        print(f"   {col}: {len(outliers)} outliers")

# 11️⃣ Inflation Rate summary
if 'Inflation Rate (%)' in df.columns:
    print("\n11️⃣ Inflation Summary:\n", df['Inflation Rate (%)'].describe())

# 12️⃣ Average unemployment rate
if 'Unemployment Rate (%)' in df.columns:
    print("\n12️⃣ Average Unemployment Rate:",
          round(df['Unemployment Rate (%)'].mean(), 2))

# 13️⃣ Index with highest trading volume
if 'Stock Index' in df.columns and 'Trading Volume' in df.columns:
    highest_vol = df.groupby('Stock Index')['Trading Volume'].mean().idxmax()
    print("\n13️⃣ Highest Trading Volume Index:", highest_vol)

# 14️⃣ Record count per index
if 'Stock Index' in df.columns:
    print("\n14️⃣ Records per Index:\n", df['Stock Index'].value_counts())

# 15️⃣ Correlation inflation vs interest
if 'Inflation Rate (%)' in df.columns and 'Interest Rate (%)' in df.columns:
    corr = df['Inflation Rate (%)'].corr(df['Interest Rate (%)'])
    print("\n15️⃣ Correlation (Inflation vs Interest):", round(corr, 3))

# 16️⃣ Average Consumer Confidence
if 'Consumer Confidence Index' in df.columns:
    print("\n16️⃣ Average Consumer Confidence Index:",
          round(df['Consumer Confidence Index'].mean(), 2))

# 17️⃣ Column with highest standard deviation
print("\n17️⃣ Column with Highest Standard Deviation:",
      numeric_df.std().idxmax())

# 18️⃣ Highest Gold Price
if 'Gold Price (USD per Ounce)' in df.columns:
    print("\n18️⃣ Highest Gold Price:",
          df['Gold Price (USD per Ounce)'].max())

# 19️⃣ Highest Crude Oil Price
if 'Crude Oil Price (USD per Barrel)' in df.columns:
    print("\n19️⃣ Highest Crude Oil Price:",
          df['Crude Oil Price (USD per Barrel)'].max())

# 20️⃣ Average Corporate Profit
if 'Corporate Profits (Billion USD)' in df.columns:
    print("\n20️⃣ Average Corporate Profit:",
          round(df['Corporate Profits (Billion USD)'].mean(), 2))

# ============================================================
print("\n✅ All 20 EDA questions answered successfully!")
```

```python
import pandas as pd

# Load your dataset
df = pd.read_csv("/content/Retail Data11.csv")

# 1. View structure
print(df.info())
print(df.isnull().sum())

# 2. Shape
print("Shape:", df.shape)

# 3. Duplicates
print("Duplicate rows:", df.duplicated().sum())

# 4. Missing or corrupted dates/numbers
for col in ['Order Date', 'Ship Date']:
    if col in df.columns:
        print(col, "missing:", df[col].isnull().sum())

# 5. Convert dates
for col in ['Order Date', 'Ship Date']:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# 6. Future or inconsistent shipping dates
if {'Order Date','Ship Date'}.issubset(df.columns):
    print("Inconsistent Ship Dates:", (df['Ship Date'] < df['Order Date']).sum())

# 7. Convert price-related columns safely
price_cols = [c for c in df.columns if any(x in c.lower() for x in ['price','cost','total','profit','discount'])]
for c in price_cols:
    df[c] = df[c].replace('[\$,]', '', regex=True).astype(float)
print("Converted:", price_cols)

# 8. Unique values
for c in ['Customer Type','Order Priority']:
    if c in df.columns:
        print(c, ":", df[c].unique())

# 9. Most common ship mode
if 'Ship Mode' in df.columns:
    print(df['Ship Mode'].value_counts())

# 10. Top cities by orders
if 'City' in df.columns:
    print(df['City'].value_counts().head())

# 11. Range of quantity & price
if 'Order Quantity' in df.columns:
    print("Order Quantity range:", df['Order Quantity'].min(), "-", df['Order Quantity'].max())

# 12. Shipping duration
if {'Order Date','Ship Date'}.issubset(df.columns):
    df['Ship Days'] = (df['Ship Date'] - df['Order Date']).dt.days
    print("Shipping Duration column added.")

# 13. Zero/Negative values
for c in ['Order Quantity','Total']:
    if c in df.columns:
        print(c, "zero or negative:", (df[c] <= 0).sum())

# 16. Top 5 products
if 'Product Name' in df.columns:
    print(df.groupby('Product Name')['Order Quantity'].sum().sort_values(ascending=False).head())

# 17. Account Manager revenue
if 'Account Manager' in df.columns and 'Total' in df.columns:
    print(df.groupby('Account Manager')['Total'].sum().sort_values(ascending=False).head(1))

# 18. Avg shipping cost by mode
if 'Ship Mode' in df.columns and 'Shipping Cost' in df.columns:
    print(df.groupby('Ship Mode')['Shipping Cost'].mean())

# 19. Most profitable product
if 'Product Name' in df.columns and 'Profit' in df.columns:
    print(df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(1))

# ===== Insights =====
if 'Total' in df.columns:
    print("Total Revenue:", df['Total'].sum())
if 'Customer Type' in df.columns and 'Total' in df.columns:
    print(df.groupby('Customer Type')['Total'].sum())
if 'Order Priority' in df.columns and 'Total' in df.columns:
    print(df.groupby('Order Priority')['Total'].sum())

```
